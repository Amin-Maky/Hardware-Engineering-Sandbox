  1. متغیر `image_folder` آدرس پوشه پروژه (`HW-1`) را نگه می‌دارد.
  2. تابع `os.listdir(image_folder)` نام تمامی فایل‌ها و پوشه‌های موجود در آن آدرس را استخراج کرده و به صورت یک لیست برمی‌گرداند که با دستور `print` چاپ می‌شود.
- **معنای خروجی:** خروجی یک لیست پایتون است که شامل نام ۴ فایل موجود در پوشه (`01_Read_Write_Display_OpenCV.ipynb`، تصویر اصلی `Me.jpg`، تصویر تازه ذخیره‌شده `SaveMe.jpg` و فایل نوت‌بوک دوم) است و تایید می‌کند که فایل `SaveMe.jpg` با موفقیت در پوشه ایجاد شده است.


```python
# Read All Image in a Folder
for filename in os.listdir(image_folder):
    
    image_path = os.path.join(image_folder, filename)
    
    image = cv2.imread(image_path)
    if image is not None:
        print(filename, image.shape)

    # print(filename, image.shape)
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** در این حلقه، تمامی فایل‌های موجود در پوشه (`os.listdir`) پیمایش شده و با `os.path.join` مسیر کامل هر فایل ساخته می‌شود. از آنجایی که در پوشه فایل‌های غیرتصویری (مانند فایل‌های `.ipynb`) نیز وجود دارند که `cv2.imread` برای آن‌ها مقدار `None` برمی‌گرداند، شرط `if image is not None:` اضافه شده است تا از بروز خطای `AttributeError` جلوگیری شود و فقط برای فایل‌هایی که واقعاً تصویر هستند و با موفقیت خوانده شده‌اند، نام و ابعادشان (`image.shape`) چاپ شود.
- **معنای خروجی:** نام دو فایل تصویری موجود در پوشه (`Me.jpg` و `SaveMe.jpg`) به همراه ابعاد ماتریس آن‌ها `(736, 736, 3)` (۷۳۶ پیکسل ارتفاع، ۷۳۶ پیکسل عرض و ۳ کانال رنگی) در خروجی چاپ شده و فایل‌های نوت‌بوک نادیده گرفته شده‌اند.


```python
# Read Just Image files in a folder
image_extensions = [".jpg", ".jpeg", ".png", ".bmp"]

for filename in os.listdir(image_folder):
    
    extension = os.path.splitext(filename)[1].lower()
    
    if extension in image_extensions:
        
        image_path = os.path.join(image_folder, filename)
        
        image = cv2.imread(image_path)
        
        print(filename, image.shape)
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** این بلوک روش اصولی‌تر برای فیلتر کردن فایل‌های تصویری قبل از خواندن آن‌ها را نشان می‌دهد:
  1. ابتدا لیستی از پسوندهای مجاز تصویری در `image_extensions` تعریف شده است.
  2. درون حلقه، تابع `os.path.splitext(filename)[1].lower()` پسوند هر فایل را جدا کرده و به حروف کوچک تبدیل می‌کند.
  3. با شرط `if extension in image_extensions` بررسی می‌شود که آیا پسوند فایل جزو پسوندهای تصویری هست یا خیر؛ و فقط در صورت تایید، تصویر با `cv2.imread` خوانده شده و نام و ابعادش چاپ می‌شود.
- **معنای خروجی:** همانند بلوک قبل، فقط فایل‌های تصویری (`Me.jpg` و `SaveMe.jpg`) شناسایی و خوانده شده و ابعاد آن‌ها `(736, 736, 3)` در خروجی نمایش داده شده است.


# Image as Numpy Array

```python
image = cv2.imread(image_path)
print(type(image))
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** در این سلول پس از خواندن تصویر با `cv2.imread`، از تابع داخلی `type(image)` استفاده شده است تا نوع داده (Data Type) شیء تصویر در پایتون مشخص شود.
- **معنای خروجی:** عبارت `<class 'numpy.ndarray'>` نشان می‌دهد که کتابخانه OpenCV تصاویر را در قالب **آرایه‌های چندبعدی نامپای (NumPy N-dimensional Array)** در حافظه ذخیره و مدیریت می‌کند؛ بنابراین تمامی عملیات‌های ماتریسی NumPy روی تصاویر قابل اجرا هستند.


```python
height, width, channels = image.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** ویژگی `image.shape` یک تاپل ۳ عضوی شامل ابعاد ماتریس تصویر برمی‌گرداند. در این کد با روش Unpacking، این سه مقدار به ترتیب در متغیرهای `height` (ارتفاع/تعداد سطرها)، `width` (عرض/تعداد ستون‌ها) و `channels` (تعداد کانال‌های رنگی) ریخته شده و چاپ می‌شوند.
- **معنای خروجی:** خروجی به تفکیک نشان می‌دهد که تصویر دارای ارتفاع **736** پیکسل، عرض **736** پیکسل و **3** کانال رنگی (آبی، سبز و قرمز - BGR) است.


```python
pixel = image[100, 200]
print(pixel)
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** با استفاده از اندیس‌گذاری آرایه به صورت `image[100, 200]`، به پیکسل واقع در **سطر ۱۰۰** (محور عمودی $y$) و **ستون ۲۰۰** (محور افقی $x$) دسترسی پیدا کرده و مقادیر رنگی آن را در متغیر `pixel` ذخیره و چاپ می‌کنیم.
- **معنای خروجی:** آرایه `[126  87  79]` شدت روشنایی سه کانال رنگی در آن پیکسل خاص را به ترتیب فرمت **BGR** نشان می‌دهد (یعنی شدت آبی = ۱۲۶، سبز = ۸۷ و قرمز = ۷۹).


```python
print(pixel[0])
print(pixel[1])
print(pixel[2])
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** در این سلول با اندیس‌گذاری روی متغیر `pixel` (از اندیس `0` تا `2`)، مقادیر هر یک از کانال‌های رنگیِ آن پیکسل به صورت مجزا استخراج و در خطوط جداگانه چاپ شده‌اند.
- **معنای خروجی:** اعداد چاپ‌شده به ترتیب بیانگر شدت کانال آبی (`pixel[0] = 126`)، کانال سبز (`pixel[1] = 87`) و کانال قرمز (`pixel[2] = 79`) در پیکسلِ `(100, 200)` هستند.


```python
print(image.dtype)
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** ویژگی `image.dtype` نوع داده (Data Type) عناصر تشکیل‌دهنده آرایه نامپای (مقادیر پیکسل‌ها) را مشخص می‌کند.
- **معنای خروجی:** عبارت `uint8` به معنای **عدد صحیح ۸ بیتی بدون علامت (Unsigned 8-bit Integer)** است. این یعنی مقدار هر کانال رنگی در هر پیکسل یک عدد صحیح در بازه $0$ تا $255$ ($2^8 - 1$) است که $0$ نشان‌دهنده تاریکی مطلق و $255$ نشان‌دهنده بیشترین شدت روشنایی آن رنگ است.

- **معنای خروجی:** خروجی، ناحیه برش‌خورده از تصویر اصلی با ابعاد ۴۰۰ در ۵۰۰ پیکسل است که دقیقاً روی بخش میانی/بالایی تصویر (گربه خاکستری در پس‌زمینه) زوم کرده است.


### Change One pixel color

**توضیحات کد و خروجی:**

- **معنای خروجی:** در تصویر خروجی، دقیقاً در مختصات `(y=100, x=200)` یک تک‌پیکسل به رنگ قرمز درآمده است (که به دلیل رزولوشن بالای ۷۳۶×۷۳۶ تصویر و کوچک بودنِ تنها یک پیکسل، به صورت یک نقطه بسیار ریز قرمز در بالای سر گربه قابل مشاهده است).


### Change color of a region

```python
image[100:300, 200:400] = [0, 0, 255]
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.show()
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** در این سلول، مقدار پیکسل واقع در سطر ۱۰۰ و ستون ۲۰۰ مستقیماً به آرایه `[0, 0, 255]` تغییر داده شده است. از آنجا که تصویر اصلی در قالب BGR است، مقداردهی `(B=0, G=0, R=255)` به معنای **رنگ قرمز خالص** است. سپس تصویر به RGB تبدیل شده و نمایش داده می‌شود.
- **معنای خروجی:** در تصویر خروجی، دقیقاً در مختصات `(y=100, x=200)` یک تک‌پیکسل به رنگ قرمز درآمده است (که به دلیل رزولوشن بالای ۷۳۶×۷۳۶ تصویر و کوچک بودنِ تنها یک پیکسل، به صورت یک نقطه بسیار ریز قرمز در بالای سر گربه قابل مشاهده است).


```python
image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

crop = image[100:500, 200:700]
crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

image[100:300, 200:400] = [0, 0, 255]
image_rgb_Changed = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(crop_rgb)
plt.title("Crop")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(image_rgb_Changed)
plt.title("image_rgb_Changed")
plt.axis("off")

plt.show()
```

**توضیحات کد و خروجی:**
- **کارکرد کد:** 
  1. ابتدا تصویر اصلی دوباره خوانده شده و به فضای رنگی RGB تبدیل می‌شود (`image_rgb`).
  2. بخشی از تصویر (سطرهای ۱۰۰ تا ۵۰۰ و ستون‌های ۲۰۰ تا ۷۰۰) برش داده شده و به RGB تبدیل می‌گردد (`crop_rgb`).
  3. با دستور `image[100:300, 200:400] = [0, 0, 255]`، به جای یک پیکسل، یک **ناحیه مستطیلی/مربعی ۲۰۰×۲۰۰ پیکسلی** (از سطر ۱۰۰ تا ۳۰۰ و ستون ۲۰۰ تا ۴۰۰) در تصویر اصلی با رنگ قرمز خالص `(B=0, G=0, R=255)` مقداردهی می‌شود و سپس به RGB تبدیل می‌گردد (`image_rgb_Changed`).
  4. در نهایت با استفاده از `plt.figure(figsize=(10, 5))` یک بوم با ابعاد ۱۰×۵ اینچ ساخته شده و با دستور `plt.subplot(1, 3, i)` خروجی به یک جدول **۱ سطری و ۳ ستونی** تقسیم می‌شود تا هر سه تصویر در کنار هم ترسیم شوند.
- **معنای خروجی:** در خروجی سه تصویر به صورت مقایسه‌ای در یک ردیف نمایش داده شده‌اند:
  - **سمت چپ (`Original`):** تصویر اصلی بدون تغییر.
  - **وسط (`Crop`):** ناحیه برش‌خورده از تصویر.
  - **سمت راست (`image_rgb_Changed`):** تصویر اصلی که یک مربع قرمز رنگ (به ابعاد ۲۰۰ در ۲۰۰ پیکسل) روی ناحیه مشخص‌شده آن اعمال شده است.
